# 1. Révision générale – Semaines 1 à 7

## 1.1 Introduction au TNS (Semaine 1)

- **Définition** : traitement de signaux discrets par algorithmes.
- **Classification** : analogique, échantillonné, quantifié, numérique.
- **Applications** : audio, télécoms, imagerie, radar, compression, contrôle.
- **Avantages** : reproductibilité, immunité au bruit, flexibilité logicielle.

## 1.2 Fondements mathématiques (Semaine 2)

- **Nombre complexe** : $z = a + jb = r e^{j\theta}$, avec $r = \sqrt{a^2+b^2}$, $\theta = \arctan(b/a)$.
- **Formule d'Euler** : $e^{j\theta} = \cos\theta + j\sin\theta$.
- **Phaseur** : $A e^{j\phi}$ pour $A\cos(\omega t + \phi)$.

## 1.3 Théorie de l'échantillonnage (Semaine 3)

- **Théorème de Nyquist** : $f_s > 2 f_{\max}$.
- **Repliement (aliasing)** : si $f_s < 2f_{\max}$ → distorsion irréversible.
- **Quantification** : erreur maximale $\Delta/2$.
- **CAN** : échantillonnage → quantification → codage.

## 1.4 Signaux discrets et systèmes LTI (Semaine 4)

- **Impulsion** : $\delta[n]$ ; **échelon** : $u[n]$.
- **Convolution** : $y[n] = \sum_k x[k] h[n-k]$.
- **Stabilité BIBO** : $\sum |h[n]| < \infty$.
- **Causalité** : $h[n] = 0$ pour $n < 0$.

## 1.5 Transformée en Z (Semaine 5)

- $X(z) = \sum x[n] z^{-n}$ ; **ROC** : région de convergence.
- **Pôles** = annulent le dénominateur ; **zéros** = annulent le numérateur.
- **Stabilité** (système causal) : tous les pôles à l'intérieur du cercle unité ($|z|<1$).

## 1.6 Domaine fréquentiel (Semaine 6)

- **Série de Fourier** : décomposition d'un signal périodique en harmoniques.
- **DTFT** : $X(e^{j\omega}) = \sum x[n] e^{-j\omega n}$, périodique de période $2\pi$.
- **Relation TZ ↔ DTFT** : $X(e^{j\omega}) = X(z)|_{z=e^{j\omega}}$.
- **Parseval** : $\sum |x[n]|^2 = \frac{1}{2\pi}\int_{-\pi}^{\pi} |X(e^{j\omega})|^2 d\omega$.

# 2. Révision approfondie – Semaines 8 à 13 (accent examen)

## 2.1 DFT et FFT (Semaine 8)

- **DFT** : $X[k] = \sum_{n=0}^{N-1} x[n] e^{-j2\pi kn/N}$, $k = 0, \dots, N-1$.
- **FFT** : algorithme rapide $O(N\log N)$ (Cooley-Tukey).
- **Résolution fréquentielle** : $\Delta f = f_s/N$.
- **Fuite spectrale** : due à la troncature du signal (fenêtre rectangulaire implicite).
- **Fenêtrage** : Hamming, Hanning, Blackman, Flat-top.
  - Compromis : résolution ↔ dynamique.
  - Règle : Hamming = bon compromis général.
- **Zero-padding** : interpole le spectre, n'améliore **pas** la résolution réelle.
- **Parseval (DFT)** : $\sum |x[n]|^2 = \frac{1}{N}\sum |X[k]|^2$.

## 2.2 Filtres numériques RIF/RII (Semaine 9)

### Filtres RIF (FIR)
- **Équation** : $y[n] = \sum_{k=0}^{M} b_k x[n-k]$.
- **Phase linéaire** si coefficients symétriques.
- **Toujours stables**.
- **Conception** : méthode de la fenêtre, échantillonnage en fréquence, Parks-McClellan.
- **Ordre élevé** pour pente raide.

### Filtres RII (IIR)
- **Équation** : $y[n] = \sum b_k x[n-k] - \sum a_k y[n-k]$.
- **Pente raide** avec ordre faible.
- **Phase non linéaire**.
- **Stabilité** conditionnelle (pôles dans cercle unité).
- **Conception** : Butterworth, Chebyshev, elliptique.
- **Implémentation** : forme directe I/II, cascade de biquads.

### Comparaison

| Critère | RIF | RII |
|---|---|---|
| Stabilité | Toujours stable | Conditionnelle |
| Phase | Linéaire | Non linéaire |
| Ordre | Élevé | Faible |
| Coût CPU | Élevé | Faible |
| Application | Audio (phase) | Temps réel (efficacité) |

## 2.3 Implémentation et analyse (Semaine 10)

- **Forme directe I** : calcul direct de l'équation aux différences.
- **Forme directe II** : minimise les états (retards communs).
- **Forme transposée** : meilleure robustesse numérique.
- **Cascade de biquads** : standard pour les IIR d'ordre élevé.
- **Quantification des coefficients** : impact sur la réponse.
- **Bruit d'arrondi** : à minimiser (formats Q15/Q31).

## 2.4 TNS embarqué et CMSIS-DSP (Semaine 11)

- **CMSIS-DSP** : bibliothèque optimisée ARM pour Cortex-M.
- **Formats** : `_f32` (float), `_q15` (int16), `_q31` (int32).
- **Fonctions clés** :
  - `arm_fir_init_f32`, `arm_fir_f32` : filtrage FIR.
  - `arm_biquad_cascade_df1_f32` : filtrage IIR.
  - `arm_rfft_fast_f32` : FFT réelle.
  - `arm_cmplx_mag_f32` : module complexe.
- **Traitement par blocs** : compatible DMA + ISR.
- **Q15** : plage [−1, 1[, résolution $2^{-15}$.

## 2.5 Évaluation pratique (Semaines 12–13)

- **Partie 1** : génération, échantillonnage, FFT, filtres RIF/RII (Python).

## 5.5 Exemple de projet : détection de notes de musique

Ce projet illustre une chaîne TNS complète : **génération → analyse → filtrage → décision**.

```python
# --- Exemple de projet : détection de notes de musique ---
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq
from scipy.signal import find_peaks, butter, filtfilt

# Paramètres
fs = 8000
duration = 1.0
t = np.linspace(0, duration, int(fs*duration), endpoint=False)

# Notes jouées (gamme de Do majeur)
notes = {
    'Do4': 261.63, 'Ré4': 293.66, 'Mi4': 329.63,
    'Fa4': 349.23, 'Sol4': 392.00, 'La4': 440.00, 'Si4': 493.88, 'Do5': 523.25
}

# Signal test : 3 notes successives
notes_jouees = ['Do4', 'Mi4', 'Sol4']
x = np.zeros_like(t)
segment = len(t) // len(notes_jouees)
for i, note in enumerate(notes_jouees):
    f0 = notes[note]
    seg = slice(i*segment, (i+1)*segment)
    x[seg] = np.sin(2*np.pi*f0*t[seg]) + 0.5*np.sin(2*np.pi*2*f0*t[seg]) + 0.3*np.sin(2*np.pi*3*f0*t[seg])

# Bruit léger
x += 0.02 * np.random.randn(len(x))

# 1. Analyse spectrale globale
N = len(x)
window = np.hamming(N)
X = fft(x * window)
freqs = fftfreq(N, 1/fs)[:N//2]
mag = 2*np.abs(X[:N//2]) / np.sum(window)

# 2. Analyse par segments (STFT simplifiée)
n_segments = len(notes_jouees)
taille_seg = len(x) // n_segments
notes_detectees = []

for i in range(n_segments):
    seg = x[i*taille_seg:(i+1)*taille_seg]
    w = np.hamming(len(seg))
    X_seg = fft(seg * w)
    freqs_seg = fftfreq(len(seg), 1/fs)[:len(seg)//2]
    mag_seg = 2*np.abs(X_seg[:len(seg)//2]) / np.sum(w)
    # Détection de la fréquence dominante
    peaks, props = find_peaks(mag_seg, height=0.1)
    if len(peaks) > 0:
        f_dom = freqs_seg[peaks[np.argmax(props['peak_heights'])]]
        # Identification de la note la plus proche
        note_detectee = min(notes, key=lambda n: abs(notes[n] - f_dom))
        notes_detectees.append((note_detectee, f_dom))

print("Notes jouées    :", notes_jouees)
print("Notes détectées :", [n for n, _ in notes_detectees])
print("Fréquences      :", [f"{f:.1f} Hz" for _, f in notes_detectees])

# Visualisation
fig, axes = plt.subplots(2, 1, figsize=(14, 8))
axes[0].plot(t, x)
axes[0].set_title('Signal audio (3 notes successives)')
axes[0].set_xlabel('Temps (s)')
axes[0].set_ylabel('Amplitude')

axes[1].plot(freqs, mag)
axes[1].set_xlim(0, 1000)
axes[1].set_xlabel('Fréquence (Hz)')
axes[1].set_ylabel('Magnitude')
axes[1].set_title('Spectre global (3 notes)')
for note in notes_jouees:
    axes[1].axvline(notes[note], color='r', linestyle='--', alpha=0.5)
    axes[1].text(notes[note], max(mag)*0.9, note, rotation=90, va='top')

plt.tight_layout()
plt.show()
```

## Formulaire à mémoriser

| Notion | Formule |
|---|---|
| Nyquist | $f_s > 2f_{\max}$ |
| Résolution DFT | $\Delta f = f_s/N$ |
| Convolution | $y[n] = \sum x[k]h[n-k]$ |
| Transformée en Z | $X(z) = \sum x[n]z^{-n}$ |
| DTFT | $X(e^{j\omega}) = X(z)|_{z=e^{j\omega}}$ |
| Parseval | $\sum\|x[n]\|^2 = \frac{1}{2\pi}\int\|X(e^{j\omega})\|^2 d\omega$ |
| BIBO | $\sum\|h[n]\| < \infty$ |
| Stabilité | $\|z_i\| < 1$ |
| Q15 | $\Delta = 2^{-15}$ |
| SNR (dB) | $10\log_{10}(P_s/P_b)$ |
| Complexité FFT | $O(N\log N)$ |
| Complexité DFT | $O(N^2)$ |
