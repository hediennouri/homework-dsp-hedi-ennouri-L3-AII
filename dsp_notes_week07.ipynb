# 1. Rappels théoriques des semaines 1 à 6

## 1.1 Introduction au TNS (Semaine 1)

- **Définition** : le TNS est l'ensemble des techniques mathématiques et algorithmiques permettant d'analyser, modifier et synthétiser des signaux discrets.
- **Classification des signaux** :
  - *Analogique* : continu en temps et en amplitude.
  - *Échantillonné* : discret en temps, continu en amplitude.
  - *Quantifié* : continu en temps, discret en amplitude.
  - *Numérique* : discret en temps et en amplitude.
- **Applications** : audio, télécoms, imagerie médicale, radar, compression, contrôle industriel.

## 1.2 Fondements mathématiques (Semaine 2)

- **Nombre complexe** : $z = a + jb = r e^{j\theta}$ avec $r = \sqrt{a^2+b^2}$ et $\theta = \arctan(b/a)$.
- **Formule d'Euler** : $e^{j\theta} = \cos\theta + j\sin\theta$.
- **Sinusoïde** : $x(t) = A \cos(2\pi f t + \phi)$.
- **Phaseur** : représentation complexe $X = A e^{j\phi}$.

## 1.3 Théorie de l'échantillonnage (Semaine 3)

- **Théorème de Nyquist-Shannon** : un signal de bande passante $B$ peut être reconstruit exactement si $f_s > 2B$.
- **Fréquence de Nyquist** : $f_N = f_s/2$.
- **Repliement de spectre (aliasing)** : si $f_s < 2B$, les hautes fréquences se replient dans la bande utile.
- **Quantification** : discrétisation de l'amplitude → erreur de quantification $\Delta = q/2$ (q = pas de quantification).
- **CAN** : échantillonnage → quantification → codage.

## 1.4 Signaux à temps discret et systèmes LTI (Semaine 4)

- **Impulsion unité** : $\delta[n] = 1$ si $n=0$, $0$ sinon.
- **Échelon unité** : $u[n] = 1$ si $n \ge 0$, $0$ sinon.
- **Système LTI** : linéaire et invariant dans le temps.
- **Convolution** : $y[n] = x[n] * h[n] = \sum_{k=-\infty}^{+\infty} x[k] h[n-k]$.
- **Réponse impulsionnelle** : $h[n]$ est la sortie du système lorsque l'entrée est $\delta[n]$.
- **Stabilité BIBO** : $\sum_{n=-\infty}^{+\infty} |h[n]| < \infty$.

## 1.5 Transformée en Z (Semaine 5)

- **Définition** : $X(z) = \sum_{n=-\infty}^{+\infty} x[n] z^{-n}$.
- **ROC** : région du plan complexe où la série converge.
- **Pôles et zéros** : valeurs de $z$ qui annulent respectivement le dénominateur et le numérateur.
- **Stabilité** : un système causal est stable si tous ses pôles sont à l'intérieur du cercle unité ($|z| < 1$).

## 1.6 Domaine fréquentiel (Semaine 6)

- **Série de Fourier** : décomposition d'un signal périodique en harmoniques.
- **DTFT** : $X(e^{j\omega}) = \sum_{n=-\infty}^{+\infty} x[n] e^{-j\omega n}$, périodique de période $2\pi$.
- **Relation avec la transformée en Z** : $X(e^{j\omega}) = X(z)|_{z=e^{j\omega}}$.
- **Spectre d'amplitude et de phase** : $|X(e^{j\omega})|$ et $\angle X(e^{j\omega})$.
- **Parseval** : $\sum |x[n]|^2 = \frac{1}{2\pi} \int_{-\pi}^{\pi} |X(e^{j\omega})|^2 d\omega$.

## Exercice 1 – Calcul de transformée en Z

Soit $x[n] = (0.5)^n u[n]$. Calculez sa transformée en Z et précisez sa ROC.

**Solution** :
$$X(z) = \sum_{n=0}^{\infty} (0.5)^n z^{-n} = \frac{1}{1 - 0.5 z^{-1}}, \quad |z| > 0.5$$

Vérification en Python :

```python
import numpy as np
import matplotlib.pyplot as plt

# Vérification numérique de la ROC
z = np.linspace(0.1, 2, 1000)
X = 1 / (1 - 0.5 / z)  # z^{-1} = 1/z
plt.plot(z, np.abs(X))
plt.axvline(0.5, color='r', linestyle='--', label='|z| = 0.5')
plt.xlabel('|z|')
plt.ylabel('|X(z)|')
plt.title('ROC de x[n] = (0.5)^n u[n]')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Exercice 2 – Échantillonnage et repliement

Un signal $x(t) = \cos(2\pi \cdot 100 t)$ est échantillonné à $f_s = 150$ Hz.

1. Quelle est la fréquence de Nyquist ?
2. Y a-t-il repliement ? Si oui, quelle fréquence apparente observe-t-on ?

**Solution** :
- $f_N = f_s/2 = 75$ Hz.
- Le signal a une fréquence de 100 Hz > 75 Hz, donc il y a repliement.
- La fréquence apparente est $|100 - 150| = 50$ Hz.

Vérification en Python :

```python
fs = 150
t = np.linspace(0, 0.1, 1000)
x = np.cos(2*np.pi*100*t)

# Échantillonnage
n = np.arange(0, 0.1, 1/fs)
x_ech = np.cos(2*np.pi*100*n)

plt.plot(t, x, label='Signal original (100 Hz)')
plt.stem(n, x_ech, linefmt='r-', markerfmt='ro', basefmt=' ', label='Échantillons')
plt.plot(t, np.cos(2*np.pi*50*t), 'g--', label='Signal apparent (50 Hz)')
plt.xlabel('Temps (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(alpha=0.3)
plt.title('Repliement de spectre : 100 Hz échantillonné à 150 Hz')
plt.show()
```
